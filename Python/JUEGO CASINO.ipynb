```python
# from random import randint
from random import shuffle



#  DATOS DEL JUEGO


# Partidos disponibles: diccionario donde cada deporte tiene una lista de tuplas (equipo local, equipo visitante)

PARTIDOS = {
    "Fútbol": [
        ("Real Madrid", "FC Barcelona"),
        ("Atlético de Madrid", "Sevilla"),
        ("Valencia", "Athletic Club"),
        ("Las Palmas", "Almeria"),],
    
    "Baloncesto": [
        ("Real Madrid Baloncesto", "Barça Basket"),
        ("Valencia Basket", "Unicaja"),
        ("Baskonia", "Joventut Badalona"),
        ("Houston Rockets", "Dallas Mavericks"),
        ("San Antonio Spurs", "Golden State Warriors"),],
}

# Tipos de apuesta: lista de tuplas nombre de la apuesta, cuota.
# La cuota es lo que multiplica a lo apostado si se gana
TIPOS_APUESTA = {
    "Fútbol": [
        ("Local", 2.0),
        ("Empate", 3.0),
        ("Visitante", 2.5),
        ("+2.5 goles", 1.8),
        ("Ambos marcan", 1.9),
    ],
    "Baloncesto": [
        ("Local", 1.9),
        ("Visitante", 2.1),
        ("+200.5 puntos", 1.8),
        ("Local gana por 10 o más", 2.8),
        ("Ambos superan los 100 puntos", 2.2),
    ],
}

# Números rojos de la ruleta (los demás son negros  salvo el 0)
NUMEROS_ROJOS = [1, 3, 5, 7, 9, 12, 14, 16, 18, 19, 21, 23, 25, 27, 30, 32, 34, 36]




def pedir_numero(mensaje, minimo, maximo):
    #"""Pide un número entero al jugador hasta que sea válido
    #y esté entre minimo y maximo (ambos incluidos)."""
    while True:
        texto = input(mensaje)

# isdigit() nos dice si el texto está formado solo por números
        if texto.isdigit() == False:
            print("Introduce un número entero válido.")
            continue

        numero = int(texto)

        if numero < minimo or numero > maximo:
            print(f"El número debe estar entre {minimo} y {maximo}.")
            continue

        return numero


def preguntar_si_no(mensaje):
    #Pregunta algo al jugador. Devuelve True si responde 'si'
    #y False si responde 'no'."""
    respuesta = ''
    while respuesta not in ['si', 'no']:
        respuesta = input(mensaje).lower()
        if respuesta not in ['si', 'no']:
            print("Responde con 'si' o 'no' .")

    if respuesta == 'si':
        return True
    else:
        return False


def pedir_cantidad(saldo):
   # Paso Introducir cantidad apostada.
    print(f"Tu saldo actual es: {saldo} €")
    cantidad = pedir_numero("¿Cuántos euros quieres apostar? ", 1, 1000000)
    return cantidad


def comprobar_saldo(saldo, cantidad):
    #Paso '¿Saldo suficiente?'
    #Devuelve True si hay saldo y False si no.
    if cantidad <= saldo:
        return True
    else:
        print("Saldo insuficiente")
        return False


def actualizar_saldo(saldo, ganado, ganancia, cantidad):
    #Pasos Ganar dinero / Perder apuesta / Actualizar saldo.
    #ganancia: dinero que se gana (sin contar lo apostado)
    #cantidad: dinero apostado, que se pierde si no se gana
    if ganado:
        saldo = saldo + ganancia
        print(f"¡HAS GANADO {ganancia} €!")
    else:
        saldo = saldo - cantidad
        print(f"Has perdido {cantidad} €.")

    print(f"Tu nuevo saldo es: {saldo} €")
    return saldo



#  APUESTAS DEPORTIVAS


def elegir_partido(deporte):
    #Paso Elegir partido. Devuelve una tupla local, visitante.
    lista_partidos = PARTIDOS[deporte]

    print("")
    print(f"--- PARTIDOS DE {deporte.upper()} ---")
    for i, partido in enumerate(lista_partidos):
        print(f"{i + 1}. {partido[0]} vs {partido[1]}")

    opcion = pedir_numero("Elige un partido: ", 1, len(lista_partidos))
    return lista_partidos[opcion - 1]


def elegir_tipo_apuesta(deporte):
    """Paso 'Elegir tipo de apuesta'. Devuelve una tupla (nombre, cuota)."""
    lista_tipos = TIPOS_APUESTA[deporte]

    print("")
    print("--- TIPO DE APUESTA ---")
    #lo siguiente recorre los elementos de la lista tipos
    for i, tipo in enumerate(lista_tipos):
        print(f"{i + 1}. {tipo[0]} (cuota {tipo[1]})")
#-1 porque python empieza en 0 a recorrer la lista
    opcion = pedir_numero("Elige el tipo de apuesta: ", 1, len(lista_tipos))
    return lista_tipos[opcion - 1]


def simular_futbol(partido, tipo_apuesta):
    #Simula un partido de fútbol y devuelve True si la apuesta gana.
    equipo_local = partido[0]
    equipo_visitante = partido[1]

    goles_local = randint(0, 4)
    goles_visitante = randint(0, 4)

    print("")
    print("Simulando el partido...")
    print(f"RESULTADO FINAL: {equipo_local} {goles_local} - {goles_visitante} {equipo_visitante}")

    # Comprobamos si se ha cumplido la apuesta elegida
    if tipo_apuesta == "Local":
        return goles_local > goles_visitante
    elif tipo_apuesta == "Empate":
        return goles_local == goles_visitante
    elif tipo_apuesta == "Visitante":
        return goles_visitante > goles_local
    elif tipo_apuesta == "+2.5 goles":
        return (goles_local + goles_visitante) > 2
    elif tipo_apuesta == "Ambos marcan":
        return goles_local > 0 and goles_visitante > 0
    else:
        return False


def simular_baloncesto(partido, tipo_apuesta):
    #Simula un partido de baloncesto y devuelve True si la apuesta gana.
    equipo_local = partido[0]
    equipo_visitante = partido[1]

    puntos_local = randint(75, 115)
    puntos_visitante = randint(75, 115)

    # En baloncesto si empatan, juegan prórroga
    while puntos_local == puntos_visitante:
        print("Empate! Se juega prórroga! Los aficionados se vuelven locos!")
        puntos_local = puntos_local + randint(0, 8)
        puntos_visitante = puntos_visitante + randint(0, 8)

    print("")
    print("Simulando partido...")
    print(f"RESULTADO FINAL: {equipo_local} {puntos_local} - {puntos_visitante} {equipo_visitante}")

    if tipo_apuesta == "Local":
        return puntos_local > puntos_visitante
    elif tipo_apuesta == "Visitante":
        return puntos_visitante > puntos_local
    elif tipo_apuesta == "+200.5 puntos":
        return (puntos_local + puntos_visitante) > 200
    elif tipo_apuesta == "Local gana por 10 o más":
        return (puntos_local - puntos_visitante) >= 10
    elif tipo_apuesta == "Ambos superan los 100 puntos":
        return puntos_local > 100 and puntos_visitante > 100
    else:
        return False


def apuestas_deportivas(saldo):
    # APUESTAS DEPORTIVAS
    #Devuelve el saldo actualizado al volver al menú principal.

    while True:
        # Menú deportes
        print("")
        print("=" * 40)
        print("        MENÚ DEPORTES")
        print("=" * 40)
        print("1. Fútbol")
        print("2. Baloncesto")
        print("3. Volver")

        opcion = pedir_numero("Elige una opción: ", 1, 3)

        if opcion == 3:
            return saldo

        if opcion == 1:
            deporte = "Fútbol"
        else:
            deporte = "Baloncesto"

        # Elegir partido
        partido = elegir_partido(deporte)

        #  Elegir tipo de apuesta
        tipo = elegir_tipo_apuesta(deporte)
        nombre_apuesta = tipo[0]
        cuota = tipo[1]

        # Introducir cantidad apostada
        cantidad = pedir_cantidad(saldo)

        # ¿Saldo suficiente? 
        if comprobar_saldo(saldo, cantidad) == False:
            # No: mostramos Saldo insuficiente y volvemos al menú de deportes
            continue

        # Simular partido 
        if deporte == "Fútbol":
            ha_ganado = simular_futbol(partido, nombre_apuesta)
        else:
            ha_ganado = simular_baloncesto(partido, nombre_apuesta)

        # Ha ganado? -> Ganar dinero / Perder apuesta -> Actualizar saldo el round sirve para redondear el numero
        ganancia = round(cantidad * cuota) - cantidad
        saldo = actualizar_saldo(saldo, ha_ganado, ganancia, cantidad)

        # Si te quedas sin dinero no se puede seguir apostando porque nadie tiene dinero infinito
        if saldo == 0:
            return saldo

        # ¿Seguir jugando? 
        seguir = preguntar_si_no("¿Quieres seguir apostando? (si/no): ")
        if seguir == False:
            return saldo



#  JUEGOS DEL CASINO


def jugar_ruleta(saldo):
    """Ruleta: apuesta a rojo, negro o a un número concreto."""
    print("")
    print("--- RULETA ---")
    print("1. Rojo (la apuesta se multiplica por 2)")
    print("2. Negro (la apuesta se multiplica por 2)")
    print("3. Número concreto del 0 al 36 (la apuesta se multiplica por 36)")

    eleccion = pedir_numero("¿A qué quieres apostar? ", 1, 3)

    numero_elegido = 0
    if eleccion == 3:
        numero_elegido = pedir_numero("¿A qué número (0-36)? ", 0, 36)

    cantidad = pedir_cantidad(saldo)
    if comprobar_saldo(saldo, cantidad) == False:
        return saldo

    # Giramos la ruleta
    numero_ganador = randint(0, 36)

    if numero_ganador == 0:
        color_ganador = "verde"
    elif numero_ganador in NUMEROS_ROJOS:
        color_ganador = "rojo"
    else:
        color_ganador = "negro"

    print("")
    print("La bola salta y salta hasta que...")
    print(f"Ha salido el {numero_ganador} ({color_ganador}).")

    # Comprobamos si ha ganado
    ha_ganado = False
    ganancia = 0

    if eleccion == 1 and color_ganador == "rojo":
        ha_ganado = True
        ganancia = cantidad
        print("¡¡¡MANDARINA!!!")
    elif eleccion == 2 and color_ganador == "negro":
        ha_ganado = True
        print("¡¡¡MANDARINA!!!")
        ganancia = cantidad
    elif eleccion == 3 and numero_ganador == numero_elegido:
        ha_ganado = True
        ganancia = cantidad * 35
        print("¡¡¡MANDARINA!!!")

    saldo = actualizar_saldo(saldo, ha_ganado, ganancia, cantidad)
    return saldo


def sumar_mano(cartas):
    #Suma las cartas de una mano de blackjack.
    #Los ases valen 11, pero si nos pasamos de 21 valen 1 vamos a poner que restamos 10 para poder hacelo que sino es un lio)
    total = 0
    ases = 0

    for carta in cartas:
        total = total + carta
        if carta == 11:
            ases = ases + 1

    while total > 21 and ases > 0:
        total = total - 10
        ases = ases - 1

    return total


def jugar_blackjack(saldo):
    #Blackjack sin figuras porque es una paranoia
    print("")
    print("--- BLACKJACK ---")
    print("Acércate a 21 sin pasarte. Si ganas, la apuesta se multiplica por 2.")

    cantidad = pedir_cantidad(saldo)
    if comprobar_saldo(saldo, cantidad) == False:
        return saldo

    # Reparto inicial: dos cartas para cada uno
    mano_jugador = [randint(1, 11), randint(1, 11)]
    mano_crupier = [randint(1, 11), randint(1, 11)]

    # ----- Turno del jugador -----
    turno_jugador = True
    while turno_jugador:
        total_jugador = sumar_mano(mano_jugador)
        print("")
        print(f"Tus cartas: {mano_jugador} -> suman {total_jugador}")
        print(f"Carta visible del crupier: {mano_crupier[0]}")

        if total_jugador > 21:
            print("¡Te has pasado de 21!")
            turno_jugador = False
        elif total_jugador == 21:
            print("¡21! Te plantas automáticamente.")
            turno_jugador = False
        else:
            print("1. Pedir carta")
            print("2. Plantarse")
            decision = pedir_numero("¿Qué haces? ", 1, 2)

            if decision == 1:
                mano_jugador.append(randint(1, 11))
            else:
                turno_jugador = False

    total_jugador = sumar_mano(mano_jugador)

    # ----- Turno del crupier (solo si el jugador no se ha pasado) -----
    if total_jugador <= 21:
        while sumar_mano(mano_crupier) < 17:
            mano_crupier.append(randint(1, 11))

    total_crupier = sumar_mano(mano_crupier)

    print("")
    print(f"Cartas del crupier: {mano_crupier} -> suman {total_crupier}")
    print(f"Tus cartas: {mano_jugador} -> suman {total_jugador}")

    # ----- Resultado -----
    if total_jugador > 21:
        # El jugador se pasa: pierde
        saldo = actualizar_saldo(saldo, False, 0, cantidad)
    elif total_crupier > 21:
        print("¡El crupier se ha pasado!")
        saldo = actualizar_saldo(saldo, True, cantidad, cantidad)
    elif total_jugador > total_crupier:
        saldo = actualizar_saldo(saldo, True, cantidad, cantidad)
    elif total_jugador == total_crupier:
        print("Empate: recuperas tu apuesta.")
        print(f"Tu saldo sigue siendo: {saldo} €")
    else:
        saldo = actualizar_saldo(saldo, False, 0, cantidad)

    return saldo


def jugar_trileros(saldo):
    """Trileros: encuentra la bola debajo de uno de los tres vasos."""
    print("")
    print("--- TRILEROS ---")
    print("Hay una bola escondida bajo uno de los tres vasos.")
    print("Si aciertas, ganas el doble de lo apostado (la apuesta se multiplica por 3).")

    cantidad = pedir_cantidad(saldo)
    if comprobar_saldo(saldo, cantidad) == False:
        return saldo

    # Igual que en el taller de funciones: 'O' es la bola
    vasos = [' ', 'O', ' ']
    shuffle(vasos)

    print("")
    print("El trilero mueve los vasos...")
    print("   [1]   [2]   [3]")

    eleccion = pedir_numero("¿Bajo qué vaso está la bola (1-3)? ", 1, 3)

    print("")
    print(f"Los vasos eran: [{vasos[0]}] [{vasos[1]}] [{vasos[2]}]")

    if vasos[eleccion - 1] == 'O':
        print("¡Has acertado!")
        saldo = actualizar_saldo(saldo, True, cantidad * 2, cantidad)
    else:
        print("¡Has fallado!")
        saldo = actualizar_saldo(saldo, False, 0, cantidad)

    return saldo


def jugar_dados(saldo):
    """Dados: se lanzan dos dados y se apuesta por la suma."""
    print("")
    print("--- DADOS ---")
    print("Se lanzan dos dados. ¿Qué crees que va a salir?")
    print("1. Menos de 7 (la apuesta se multiplica por 2)")
    print("2. Más de 7 (la apuesta se multiplica por 2)")
    print("3. Exactamente 7 (la apuesta se multiplica por 5)")

    eleccion = pedir_numero("Elige una opción: ", 1, 3)

    cantidad = pedir_cantidad(saldo)
    if comprobar_saldo(saldo, cantidad) == False:
        return saldo

    dado1 = randint(1, 6)
    dado2 = randint(1, 6)
    suma = dado1 + dado2

    print("")
    print("Lanzando los dados...")
    print(f"Han salido un {dado1} y un {dado2}. Suman {suma}.")

    ha_ganado = False
    ganancia = 0

    if eleccion == 1 and suma < 7:
        ha_ganado = True
        ganancia = cantidad
    elif eleccion == 2 and suma > 7:
        ha_ganado = True
        ganancia = cantidad
    elif eleccion == 3 and suma == 7:
        ha_ganado = True
        ganancia = cantidad * 4

    saldo = actualizar_saldo(saldo, ha_ganado, ganancia, cantidad)
    return saldo


def jugar_moneda(saldo):
    """Moneda: cara o cruz. Si aciertas, la apuesta se multiplica por 2."""
    print("")
    print("--- MONEDA ---")
    print("1. Cara")
    print("2. Cruz")

    eleccion = pedir_numero("Elige cara o cruz: ", 1, 2)

    cantidad = pedir_cantidad(saldo)
    if comprobar_saldo(saldo, cantidad) == False:
        return saldo

    lanzamiento = randint(1, 2)

    print("")
    print("La moneda gira en el aire...")
    if lanzamiento == 1:
        print("Ha salido CARA.")
    else:
        print("Ha salido CRUZ.")

    if eleccion == lanzamiento:
        saldo = actualizar_saldo(saldo, True, cantidad, cantidad)
    else:
        saldo = actualizar_saldo(saldo, False, 0, cantidad)

    return saldo


def casino(saldo):
    """Rama 'CASINO' del workflow: menú casino -> elegir juego.
    Después de jugar se vuelve al MENÚ PRINCIPAL."""
    print("")
    print("=" * 40)
    print("        MENÚ CASINO")
    print("=" * 40)
    print("1. Ruleta")
    print("2. Blackjack")
    print("3. Trileros")
    print("4. Dados")
    print("5. Moneda")
    print("6. Volver")

    opcion = pedir_numero("¿A qué juego quieres jugar? ", 1, 6)

    # Rombo '¿Qué juego?' del workflow
    if opcion == 1:
        saldo = jugar_ruleta(saldo)
    elif opcion == 2:
        saldo = jugar_blackjack(saldo)
    elif opcion == 3:
        saldo = jugar_trileros(saldo)
    elif opcion == 4:
        saldo = jugar_dados(saldo)
    elif opcion == 5:
        saldo = jugar_moneda(saldo)
    else:
        print("Volviendo al menú principal...")

    return saldo


# =====================================================================
#  SALIR
# =====================================================================

def simulacion_cierre(saldo):
    """Rama 'SALIR': simulación de cierre del juego."""
    if saldo > 1500: 
        pasos_cierre = [
        "Cerrando las mesas del casino...",
        "Cerrando la casa de apuestas...",
        "Apagando las luces de neón...",
        "Te vas con el/la crupier a casa.",
        ]
    else:
        pasos_cierre = [
        "Cerrando las mesas del casino...",
        "Cerrando la casa de apuestas...",
        "Apagando las luces de neón...",
        "Te vas solo/a a casa.",
    ]

    print("")
    for paso in pasos_cierre:
        print(paso)

    print("")
    print(f"Te vas a casa con {saldo} €.")
    print("¡Gracias por jugar! Recuerda: WinaMenos siempre gana.")


# =====================================================================
#  PROGRAMA PRINCIPAL  (INICIO -> MENÚ PRINCIPAL -> FIN)
# =====================================================================

# INICIO
saldo = 1000
jugando = True

print("=" * 40)
print("         BIENVENIDO/A A WINAMENOS")
print("=" * 40)
print(f"Empiezas con {saldo} €. ¡Suerte!")

while jugando:

    # Si el jugador se queda sin dinero, termina el juego
    if saldo == 0:
        print("")
        print("Te has quedado sin saldo, que noob. ¡GAME OVER!")
        simulacion_cierre(saldo)
        break

    # ----- MENÚ PRINCIPAL -----
    print("")
    print("=" * 40)
    print("            MENÚ PRINCIPAL")
    print("=" * 40)
    print(f"Saldo: {saldo} €")
    print("1. Apuestas")
    print("2. Casino")
    print("3. Salir")
    print("Recuerda debes introducir el número de elección")

    # ----- ¿Qué opción eligió el jugador? -----
    opcion = pedir_numero("Elige una opción: ", 1, 3)

    if opcion == 1:
        saldo = apuestas_deportivas(saldo)
    elif opcion == 2:
        saldo = casino(saldo)
    else:
        # SALIR -> FIN
        simulacion_cierre(saldo)
        jugando = False

# FIN
```
