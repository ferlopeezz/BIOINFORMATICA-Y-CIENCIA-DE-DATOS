## Bloque 3. Variables: ponerle nombre a un dato

*En la guía: «Variables y operadores».*

Una **variable** es una etiqueta pegada a un valor. Sirve para no repetir el dato y para que el
código se lea como una frase. En Python la asignación se escribe con un solo `=`.


````python
peso_kg = 78
talla_m = 1.72
id_paciente = "P01"

print(peso_kg)
print(id_paciente)

# Una variable se puede usar dentro de un cálculo: Python sustituye el nombre
# por su valor y luego opera. ** es la potencia.
imc = peso_kg / (talla_m ** 2)
print(imc)

# Y se puede reasignar: la etiqueta se despega de un valor y se pega a otro.
peso_kg = 80
print(peso_kg)

# OJO: el IMC calculado antes NO se recalcula solo.
print(imc)
````

> **Interpretación de resultados**
>
> `peso_kg` imprime 78 y después 80. Pero `imc` imprime `26.365...` **las dos veces**, también
> después de cambiar el peso: guarda el resultado del momento en que se ejecutó, no la fórmula.
> Para actualizarlo hay que volver a ejecutar la línea del cálculo.
>
> Esta es una de las confusiones más frecuentes al empezar, porque una hoja de cálculo sí lo
> recalcularía sola.
>
> Reglas para nombrar: tiene que empezar por letra (o guion bajo) y no puede llevar espacios; por
> costumbre, sin acentos ni signos, en minúsculas y separando palabras con guion bajo
> (`edad_paciente`, `pcr_basal`). El nombre no significa nada
> para Python: se lo das tú, para tu yo futuro.


## Bloque 4. Operadores: calcular y comparar

*En la guía: «Variables y operadores».*


````python
# Aritméticos: los de toda la vida, más la potencia (**), la división entera (//)
# y el resto (%).
print(5 + 3, 5 - 3, 5 * 3, 5 / 3)
print(5 ** 2, 7 // 2, 7 % 2)

# Comparación: preguntan algo y devuelven True o False.
# Esto es la base de todos los filtros que harás en la asignatura.
edad = 71
print(edad > 65)
print(edad < 65)
print(edad >= 71)
print(edad == 71)   # ATENCIÓN: dos signos igual para PREGUNTAR si son iguales
print(edad != 71)   # distinto de
````

````python
# El error clásico: un solo = asigna, dos == comparan.
# Si escribes edad = 65 no estás preguntando nada: estás cambiando la edad a 65.

# Lógicos: combinan condiciones. En Python se escriben con palabras.
saturacion = 93
print(edad > 65 and saturacion < 92)
print(edad > 65 or saturacion < 92)
print(not (edad > 65))
````

> **Interpretación de resultados**
>
> Los aritméticos dan `8 2 15 1.6666666666666667` y `25 3 1`.
>
> Con `edad = 71`: `True`, `False`, `True`, `True`, `False`.
>
> Con `saturacion = 93`: «edad>65 **y** saturación<92» es `False` (la segunda condición falla),
> pero «edad>65 **o** saturación<92» es `True` (basta con una).
>
> Lectura clínica: acabas de escribir un criterio de gravedad. Un algoritmo de triaje no es más que
> muchas de estas condiciones encadenadas.
>
> En R estos tres operadores se escriben `&`, `|` y `!` en lugar de `and`, `or` y `not`.
>
> Ojo: con columnas de una tabla (lo verás en el Bloque 7) no sirven `and`, `or` y `not`: se usan
> `&`, `|` y `~`, con cada condición entre paréntesis.


## Bloque 5. Funciones: qué son y cómo se leen

Una **función** es una máquina: introduces "algo", hace su trabajo y te devuelve un resultado.
Se escribe `nombre(argumentos)` y siempre lleva paréntesis.

- **nombre**: qué máquina quieres usar (`round`, `len`, `abs`...).
- **argumentos**: lo que le das para trabajar, separados por comas.
- **valor devuelto**: lo que sale. Si no lo guardas en una variable, se pierde.


````python
print(round(3.14159, 2))     # redondear a 2 decimales
print(len("neumologia"))     # cuántos caracteres tiene un texto
print("uci".upper())         # pasar a mayúsculas
print(abs(-4))               # valor absoluto

# Los argumentos se pueden dar por POSICIÓN (en el orden que espera la función)
# o por NOMBRE (más largo de escribir, pero mucho más claro).
print(round(3.14159, 2))
print(round(number=3.14159, ndigits=2))

# Guardar el resultado en vez de dejar que se pierda:
imc_redondeado = round(imc, 1)
print(imc_redondeado)

# Las funciones se anidan: la de dentro se ejecuta primero.
print(round(abs(-3.7), 0))
````

````python
# Algunas funciones no vienen sueltas, sino dentro de un MÓDULO que hay que
# importar primero. Es el equivalente a library() en R.
import statistics as st

print(st.mean([71, 58, 84]))

# Cuando no sepas qué hace una función, pregúntale a Python.
# Descomenta y ejecuta:
# help(round)
````

## Bloque 6. Listas: muchos datos en una sola caja

*En la guía: «Vectores y la tabla de datos».*

Hasta ahora cada variable guardaba UN valor. Una **lista** guarda una serie de valores.
Se escribe entre corchetes y con comas.

(En R el equivalente es el *vector*, que se construye con `c()`.)


````python
#Las listas se crean con corchetes y pueden contener cualquier tipo de dato, incluso mezclados.
edades = [71, 58, 84, 46, 67, 79, 52, 90, 63, 75]
print(edades)

# Funciones de resumen: se aplican a la lista entera de una vez.
import statistics as st
print("n        :", len(edades))
print("media    :", st.mean(edades))
print("mediana  :", st.median(edades))
print("desv.tip :", round(st.stdev(edades), 2))
print("minimo   :", min(edades))
print("maximo   :", max(edades))
print("suma     :", sum(edades))
````

````python
# Acceso por posición. ATENCIÓN: Python cuenta desde 0.
# Esta es la diferencia que más despistes provoca al pasar de R a Python.
print(edades[0])    # el PRIMERO
print(edades[1])    # el segundo
print(edades[-1])   # el último (los índices negativos cuentan desde el final)
print(edades[0:3])  # del primero al tercero: el límite derecho NO se incluye
````

````python
# Filtrado. En Python la forma idiomática es la "comprensión de lista":
# se lee casi como una frase en inglés.
mayores = [e for e in edades if e > 70]
print(mayores)
print(len(mayores))
print(sum(1 for e in edades if e > 70) / len(edades))

# Las listas de texto funcionan igual.
sexo = ["H", "M", "M", "H", "H", "M", "H", "M", "H", "M"]
print(sexo.count("H"), sexo.count("M"))

# Tabla de frecuencias sin escribir un bucle:
from collections import Counter
print(Counter(sexo))
````

````python
# Y ahora los ausentes en acción. Una lista con un None dentro no se puede promediar:
pcr_con_hueco = [184.8, 44.6, None, 12.5, 96.0]

try:
    print(st.mean(pcr_con_hueco))
except TypeError as e:
    print("ERROR:", e)

# Hay que quitarlos explícitamente antes de calcular.
limpios = [p for p in pcr_con_hueco if p is not None]
print(limpios)
print(round(st.mean(limpios), 3))
print("Ausentes:", sum(1 for p in pcr_con_hueco if p is None))
````

> **Interpretación de resultados**
>
> Sobre las 10 edades: n=10, media 68.5, mediana 69.0, desviación típica 14.07, mínimo 46, máximo 90,
> suma 685.
>
> `edades[0]` da **71** y `edades[1]` da 58. Si vienes de R, cuidado: allí `edades[1]` es el primero.
> `edades[-1]` da 75 y `edades[0:3]` da `[71, 58, 84]` (tres elementos, no cuatro: el límite derecho
> queda fuera).
>
> El filtro devuelve `[71, 84, 79, 90, 75]`, es decir 5 pacientes, el 50 %. `Counter(sexo)` da 5 y 5.
>
> Con el hueco: `st.mean` lanza un `TypeError` («can't convert type 'NoneType'...») porque no sabe
> operar con `None`. Tras quitarlo, la media de los cuatro restantes es 84.475.
>
> Regla de oro: si un cálculo se queja o devuelve un resultado raro sin motivo aparente, lo primero
> que hay que mirar es si hay ausentes en los datos. En R esto se resuelve con `na.rm = TRUE`.


## Bloque 7. La tabla de datos (DataFrame)

*En la guía: «Vectores y la tabla de datos».*

Un **DataFrame** es una tabla: varias columnas de la misma longitud puestas una al lado de otra.
Cada **fila** es un caso (un paciente) y cada **columna** una variable.
Es el formato de trabajo de toda la asignatura.

Aquí lo construimos a mano con 10 pacientes ficticios; en la Parte 2 lo leeremos ya hecho
desde un archivo.

Una diferencia con R que conviene tener clara: R trae las tablas de fábrica (`data.frame`), mientras
que Python las obtiene de una biblioteca llamada **pandas**, que hay que importar. En Colab ya está
instalada. Por convención universal se importa como `pd`.


````python
import pandas as pd

pacientes = pd.DataFrame({
    "id_paciente": ["P01","P02","P03","P04","P05","P06","P07","P08","P09","P10"],
    "sexo": ["H","M","M","H","H","M","H","M","H","M"],
    "edad": [71, 58, 84, 46, 67, 79, 52, 90, 63, 75],
    "peso_kg": [78, 64, 55, 92, 81, 60, 70, 48, 88, 67],
    "talla_m": [1.72, 1.60, 1.55, 1.80, 1.75, 1.58, 1.68, 1.50, 1.83, 1.62],
    "pcr_mg_l": [184.8, 44.6, 210.3, 12.5, 96.0, 155.2, 33.1, 240.7, 61.4, 128.9],
    "saturacion_o2": [93, 96, 88, 98, 94, 90, 97, 86, 95, 91],
    "ingreso_uci": [0, 0, 1, 0, 0, 1, 0, 1, 0, 0],
})

pacientes
````

````python
# Primer vistazo obligatorio ante cualquier tabla nueva.
print(pacientes.shape)          # (filas, columnas)
print(len(pacientes))           # solo filas
print(list(pacientes.columns))  # nombres de las columnas
print()
pacientes.info()                # estructura: cada columna con su tipo
````

````python
# Entrar en una columna concreta: tabla["columna"].
print(pacientes["edad"].tolist())
print(pacientes["edad"].mean())
print(round(pacientes["pcr_mg_l"].mean(), 2))

# Un resumen numérico de golpe:
pacientes.describe()
````

````python
# Selección por posición (.iloc) y por etiqueta (.loc).
print(pacientes.iloc[0])                       # fila 0 entera (la primera)
print()
print(pacientes.loc[0, "edad"])                # el cruce: edad del primer paciente
print()
# OJO: con .loc el límite derecho SÍ entra (filas 0, 1 y 2); con listas y .iloc, no.
print(pacientes.loc[0:2, ["id_paciente", "edad"]])
````

````python
# Filtrar filas con una condición: la comparación produce una máscara de
# True/False y la tabla se queda solo con las filas True.
print(pacientes[pacientes["edad"] > 70])
print()
print(pacientes[pacientes["saturacion_o2"] < 92][["id_paciente", "saturacion_o2"]])
print()
# Dos condiciones a la vez: con columnas se usa & (y), | (o) y ~ (no), con
# cada condición entre paréntesis. and / or aquí dan error.
print(pacientes[(pacientes["edad"] > 70) & (pacientes["saturacion_o2"] < 92)][["id_paciente", "edad", "saturacion_o2"]])
````

````python
# Crear una columna nueva. Fíjate en que el cálculo se hace sobre las 10 filas
# de una sola vez, sin ningún bucle.
pacientes["imc"] = (pacientes["peso_kg"] / pacientes["talla_m"] ** 2).round(1)
print(pacientes["imc"].tolist())
print(pacientes.shape)   # ahora hay una columna más
# (Ojo: sin .round(1), P02 daría 24.999999999999996, por cómo guarda el
# ordenador los decimales, y en el Bloque 8 caería en "Normopeso".)
````

````python
# Y la primera pregunta con forma de investigación:
# ¿los que ingresaron en UCI eran distintos de los que no?
uci = pacientes[pacientes["ingreso_uci"] == 1]
no_uci = pacientes[pacientes["ingreso_uci"] == 0]

print("Edad media  - UCI:", round(uci["edad"].mean(), 1),
      "| no UCI:", round(no_uci["edad"].mean(), 1))
print("PCR media   - UCI:", round(uci["pcr_mg_l"].mean(), 1),
      "| no UCI:", round(no_uci["pcr_mg_l"].mean(), 1))

# La forma compacta de hacer lo mismo: agrupar y resumir.
pacientes.groupby("ingreso_uci")[["edad", "pcr_mg_l", "saturacion_o2"]].mean().round(1)
````

> **Interpretación de resultados**
>
> `shape` da `(10, 8)`, y `(10, 9)` después de crear el IMC. `info()` enseña que `id_paciente` y
> `sexo` son texto (aparecen como `object`, o como `str` en las versiones más recientes de pandas) y
> el resto numéricas (`int64` las enteras, `float64` las decimales).
>
> La edad media es 68.5 y la PCR media 116.75. El filtro por edad>70 devuelve 5 filas; el de
> saturación<92 devuelve 4 pacientes (P03, P06, P08 y P10), y con las dos condiciones a la vez salen
> esos mismos 4: todos los que saturan mal tienen más de 70 años.
>
> La columna `imc` vale 26.4, 25.0, 22.9, 28.4, 26.4, 24.0, 24.8, 21.3, 26.3 y 25.5.
>
> **Y el resultado interesante:** la edad media es **84.3** en los 3 que fueron a UCI frente a
> **61.7** en los 7 que no; la PCR media, **202.1** frente a **80.2**. Los que ingresaron en UCI

>
> Eso es exactamente lo que encontrarás en la cohorte simulada de 1.800 pacientes más adelante. Acabas de hacer, con 10 filas, la versión en miniatura del análisis de todo el curso.


## Bloque 8. Condicionales: traducir una regla clínica a código

*En la guía: «Condicionales y bucles».*

`if` / `elif` / `else` ejecuta un trozo de código **solo si** se cumple una condición.

Aquí aparece la diferencia de forma más visible entre Python y R: donde R usa llaves `{}`,
Python usa la **sangría** (los espacios al principio de la línea). En Python la sangría no es
estética, es sintaxis: si te equivocas, el programa no arranca. Fíjate también en los **dos puntos**
al final de la línea del `if`.


````python
imc_paciente = 26.4

if imc_paciente < 18.5: 
    print("Bajo peso")
elif imc_paciente < 25:
    print("Normopeso")
elif imc_paciente < 30:
    print("Sobrepeso")
else:
    print("Obesidad")
````

Cómo se lee: Python evalúa las condiciones de arriba abajo y se queda con la **primera** que
resulta verdadera; las demás ni las mira. Por eso el orden importa y por eso no hace falta escribir
«entre 25 y 30»: si hemos llegado a esa línea es porque ya sabemos que no era menor de 25.


````python
# Un if solo mira UN valor. Para clasificar las 10 filas de golpe:
import numpy as np

# np.where() es el "si... entonces... si no..." aplicado a una columna entera.
# Es el equivalente de ifelse() en R.
pacientes["satura_mal"] = np.where(pacientes["saturacion_o2"] < 92, "si", "no")
print(pacientes["satura_mal"].value_counts())
````

````python
# Y pd.cut() parte una variable numérica en categorías por puntos de corte.
# Es el equivalente de cut() en R y la forma natural de crear grupos de edad,
# estadios o cualquier clasificación clínica por umbrales.
pacientes["grupo_imc"] = pd.cut(
    pacientes["imc"],
    bins=[0, 18.5, 25, 30, np.inf],
    labels=["Bajo peso", "Normopeso", "Sobrepeso", "Obesidad"],
    right=False,
)
print(pacientes["grupo_imc"].value_counts().sort_index())
````

> **Interpretación de resultados**
>
> El `if` imprime `Sobrepeso`: 26.4 no es menor que 18.5 ni que 25, pero sí es menor que 30.
>
> `satura_mal` da `no`=6 y `si`=4. `grupo_imc` da Bajo peso=0, Normopeso=4, Sobrepeso=6, Obesidad=0.
>
> El argumento `right=False` hace que los intervalos sean `[inicio, fin)`, cerrados por abajo y
> abiertos por arriba: un IMC de exactamente 25.0 cae en «Sobrepeso», igual que en la definición
> clínica.
>
> Detalle que ahorra disgustos: cuando definas cortes, decide siempre a qué lado cae el valor exacto
> del límite y déjalo escrito.


## Bloque 9. Bucles: repetir sin repetirse

*En la guía: «Condicionales y bucles».*

Un **bucle** `for` recorre una colección y ejecuta lo mismo con cada elemento.
Se lee: «para cada X dentro de esta lista, haz esto».


````python
for e in edades:
    print("Edad:", e)
````

````python
# Muy a menudo se recorre por posición para poder leer varias columnas de la
# misma fila. range(3) genera 0, 1, 2.
for i in range(3):
    print("Paciente", pacientes.loc[i, "id_paciente"],
          "- edad", pacientes.loc[i, "edad"],
          "- IMC", pacientes.loc[i, "imc"])
````

````python
# Patrón clásico: un acumulador que se va llenando dentro del bucle.
suma = 0
for e in edades:
    suma = suma + e

print(suma)
print(suma / len(edades))

# Y ahora lo importante. Ese bucle entero se escribe así:
print(sum(edades))
print(st.mean(edades))
````

> **Interpretación de resultados**
>
> El primer bucle imprime 10 líneas. El segundo, 3 líneas con id, edad e IMC de P01, P02 y P03.
> El acumulador deja `suma = 685` y `685 / 10 = 68.5`, exactamente lo mismo que devuelven
> `sum(edades)` y `st.mean(edades)`.
>
> **Mismo resultado, mucho menos código.** En Python, y también en R, la versión sin bucle es más
> corta, más fácil de leer y muchísimo más rápida cuando los datos crecen. Con 10 pacientes da igual;
> con un millón, no da igual en absoluto. Esa diferencia se mide con un cronómetro en la
> **Parte 2** y es la puerta de entrada al Big Data.
>
> Conclusión práctica: aprende a escribir bucles (los vas a leer en código de otras personas), pero
> por defecto busca primero la versión vectorizada.


## Bloque 10. Escribir tu propia función

*En la guía: «Funciones propias y errores».*

Cuando repites tres veces el mismo cálculo, conviene empaquetarlo con un nombre.
Eso es escribir una función. La estructura es siempre la misma:

```
def nombre(parametros):
    cuerpo
    return valor
```

La palabra reservada es `def`. Los **parámetros** son los huecos que rellenarás al llamarla, y
`return` marca qué devuelve. Otra vez: dos puntos al final y sangría en el cuerpo.


````python
def calcular_imc(peso, talla):
    resultado = peso / (talla ** 2)
    return round(resultado, 1)

# Llamarla es exactamente igual que llamar a round() o a len():
print(calcular_imc(78, 1.72))
print(calcular_imc(peso=48, talla=1.50))
````

````python
def clasificar_imc(imc):
    if imc is None or pd.isna(imc):
        return None
    elif imc < 18.5:
        return "Bajo peso"
    elif imc < 25:
        return "Normopeso"
    elif imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"

print(clasificar_imc(26.4))
print(clasificar_imc(17.2))
print(clasificar_imc(None))
````

Fíjate en que la primera comprobación es «¿es un ausente?». Prever el ausente desde el principio
evita la mitad de los errores en datos clínicos reales.


````python
# .apply() aplica una función a cada elemento de una columna y recoge las
# respuestas. Es la alternativa idiomática al bucle for cuando el objetivo es
# transformar, no imprimir. (En R, sapply().)
print(pacientes["imc"].apply(clasificar_imc).tolist())
print()
print(pacientes["imc"].apply(clasificar_imc).value_counts())
````

> **Interpretación de resultados**
>
> `calcular_imc(78, 1.72)` da 26.4 y `calcular_imc(48, 1.50)` da 21.3, los mismos valores que ya
> había en la columna `imc`.
>
> `clasificar_imc` devuelve `Sobrepeso`, `Bajo peso` y `None`. Aplicada a la columna entera devuelve
> las 10 clasificaciones: 6 «Sobrepeso» y 4 «Normopeso».
>
> Idea de fondo: una función es una regla escrita **una** vez y aplicada **muchas**. Si mañana
> cambian los puntos de corte, tocas un solo sitio.


## Bloque 11. El error forma parte del oficio

*En la guía: «Funciones propias y errores».*

Este bloque provoca errores **a propósito**.

Todo el mundo que programa ve mensajes rojos constantemente: no significan que hayas roto nada ni
que se te dé mal. Son la forma que tiene Python de decirte, con bastante precisión, qué no ha
entendido. Aprender a leerlos es la habilidad que más tiempo te va a ahorrar en todo el curso.

Usamos `try / except` para que el error se muestre pero el cuaderno continúe.

**Cómo se lee un error de Python.** El bloque de texto rojo se llama *traceback* y se lee
**de abajo arriba**: la última línea dice el tipo de error y el mensaje, y es casi siempre la única
que necesitas.


````python
# ERROR 1. Objeto que no existe (casi siempre, una errata en el nombre).
try:
    print(edda)
except Exception as e:
    print(type(e).__name__, "->", e)
````

````python
# ERROR 2. Función que no existe (errata en el nombre de la función).
try:
    medi(edades)
except Exception as e:
    print(type(e).__name__, "->", e)
````

````python
# ERROR 3. Tipo equivocado: sumar texto y número.
try:
    "3" + 4
except Exception as e:
    print(type(e).__name__, "->", e)
````

````python
# ERROR 4. Posición que no existe.
try:
    edades[20]
except Exception as e:
    print(type(e).__name__, "->", e)

# Compáralo con R: allí edades[20] devuelve NA SIN protestar, y el problema
# aparece tres pasos después. Python prefiere detenerse; R, seguir.
````

````python
# ERROR 5. Sangría incorrecta. Este ni siquiera llega a ejecutarse:
# Python lo rechaza al leerlo.
codigo_mal = '''
if 5 > 3:
print("hola")
'''
try:
    exec(codigo_mal)
except Exception as e:
    print(type(e).__name__, "->", e)
````

````python
# Y un aviso (Warning), que NO es un error: la orden sí se ejecuta,
# pero Python te advierte de algo.
import warnings
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter("always")
    resultado = np.mean([])          # media de una lista vacía
    print("resultado:", resultado)
    if w:
        print("AVISO:", w[0].message)
````

> **Interpretación de resultados**
>
> 1. `NameError -> name 'edda' is not defined`. Python busca ese nombre exacto y no lo encuentra.
>    Revisa la ortografía y las mayúsculas (Python distingue `edad` de `Edad`) y comprueba si
>    llegaste a ejecutar la celda que lo creaba. **En los cuadernos, además de las erratas, la causa más
>    frecuente es ejecutar las celdas en desorden.**
> 2. `NameError -> name 'medi' is not defined`: lo mismo, pero con una función. O está mal escrita,
>    o vive en un módulo que no has importado.
> 3. `TypeError -> can only concatenate str (not "int") to str`: has pedido una operación con algo
>    que no es un número. El `"3"` entre comillas es texto.
> 4. `IndexError -> list index out of range`: has pedido la posición 20 de una lista de 10.
> 5. `IndentationError -> expected an indented block after 'if' statement…`: el cuerpo del `if` tiene que ir sangrado.
> 6. La media de una lista vacía devuelve `nan` con un aviso: no es un error, es Python diciendo que
>    el resultado no es un número.
>
> **Método para depurar, en este orden:** (a) lee la ÚLTIMA línea del traceback; (b) mira el nombre
> que aparece entrecomillado; (c) ejecuta por separado los trozos de la orden para ver cuál falla;
> (d) comprueba que el objeto existe y es del tipo que crees (`type(x)`, `len(x)`, `x.shape`).

Es un **checkpoint**: una forma de que el propio código te diga si vas bien, sin tener que comparar
números a ojo. Los verás a lo largo del curso. (En R, `stopifnot()`.)


````python
assert len(pacientes) == 10, "La tabla deberia tener 10 filas"
assert pacientes.shape[1] >= 9, "Falta alguna columna"
assert round(pacientes["edad"].mean(), 1) == 68.5, "La edad media no cuadra"
assert pacientes["ingreso_uci"].sum() == 3, "Deberia haber 3 ingresos en UCI"
assert pacientes["imc"].notna().all(), "Hay algun IMC sin calcular"

print("CHECKPOINT SUPERADO: la tabla tiene 10 pacientes, 3 en UCI y el IMC calculado.")
````

## Bloque 13. Lo mismo, en R

En esta asignatura conviven R y Python. No son dos mundos distintos: son dos formas de decir las
mismas cosas. Esta tabla es tu diccionario.

| Idea | R | Python |
|---|---|---|
| Comentario | `# texto` | `# texto` |
| Asignar | `x <- 5` | `x = 5` |
| Mostrar | `print(x)` | `print(x)` |
| Verdadero / falso | `TRUE` / `FALSE` | `True` / `False` |
| Ausente | `NA` | `None` / `NaN` |
| Tipo de un objeto | `class(x)` | `type(x)` |
| Serie de valores | `c(1, 2, 3)` | `[1, 2, 3]` |
| Primer elemento | `x[1]` | `x[0]` (¡desde 0!) |
| Cuántos elementos | `length(x)` | `len(x)` |
| Media | `mean(x)` | `st.mean(lista)` / `serie.mean()` |
| Y / O / NO | `&` `\|` `!` | `and` `or` `not` (con columnas: `&` `\|` `~`) |
| Tabla de datos | `data.frame(...)` | `pd.DataFrame(...)` |
| Una columna | `tabla$edad` | `tabla["edad"]` |
| Filtrar filas | `tabla[tabla$edad > 70, ]` | `tabla[tabla["edad"] > 70]` |
| Condicional vectorizado | `ifelse(...)` | `np.where(...)` |
| Cortes en categorías | `cut(...)` | `pd.cut(...)` |
| Aplicar una función | `sapply(v, f)` | `serie.apply(f)` |
| Definir función | `f <- function(x) { }` | `def f(x):` |
| Checkpoint | `stopifnot(...)` | `assert ...` |
| Instalar / cargar | `install.packages()` / `library()` | `pip install` / `import` |

**Las dos diferencias que más despistan al principio:**

1. R cuenta desde **1** y Python desde **0**.
2. R delimita los bloques con llaves `{}` y Python con la **sangría**. En Python la sangría no es
   estética: es sintaxis, y si te equivocas el programa no arranca.

**Consejo para el curso:** elige UN lenguaje y quédate en él. El otro está aquí como espejo, para
que reconozcas el código cuando lo veas, no para que hagas todo dos veces.
