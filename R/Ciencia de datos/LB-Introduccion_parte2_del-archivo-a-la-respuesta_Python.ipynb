## Bloque 1. Bibliotecas: qué son y por qué se importan

Python viene con un conjunto de funciones de fábrica (`print`, `len`, `sum`, `round`...), pero la
mayor parte de lo útil para datos vive en **bibliotecas** que ha escrito otra gente.

## Bloque 7. El cronómetro: por qué la forma de escribir el código importa

*En la guía: «El cronómetro y la reproducibilidad».*

Este es el bloque que explica la palabra **Big Data** del temario, y la mejor forma de entenderlo
es medirlo.

En la Parte 1 viste que un bucle `for` y una función como `mean()` dan el mismo resultado. Con
10 pacientes eso es cierto y da igual cuál uses. Vamos a ver qué pasa cuando hay un millón.


```python
import time

# Primero, sobre la cohorte simulada de 1.800 pacientes.
# time.perf_counter() devuelve un instante; restando dos instantes se obtienen
# los segundos transcurridos.

inicio = time.perf_counter()
suma = 0
for i in range(len(cohorte)):
    suma += cohorte["edad"].iloc[i]
media_bucle = suma / len(cohorte)
t_bucle = time.perf_counter() - inicio

inicio = time.perf_counter()
media_vectorizada = cohorte["edad"].mean()
t_vect = time.perf_counter() - inicio

print(f"bucle        : {t_bucle:.4f} s  -> media {media_bucle:.1f}")
print(f"vectorizado  : {t_vect:.6f} s  -> media {media_vectorizada:.1f}")
print(f"el bucle tarda {t_bucle/t_vect:.0f} veces más")
```

Con 1.800 filas los dos tardan poco y la diferencia parece irrelevante.

Ahora fabricamos una tabla de **un millón de filas** remuestreando la cohorte (repetimos pacientes
al azar: no es un dato nuevo, es un banco de pruebas). La semilla fija el azar para que el resultado
sea reproducible.


```python
rng = np.random.default_rng(123)
indices = rng.integers(0, len(cohorte), 1_000_000)

grande = pd.DataFrame({
    "edad": cohorte["edad"].values[indices],
    "pcr_mg_l": cohorte["pcr_mg_l"].values[indices],
    "saturacion_o2": cohorte["saturacion_o2"].values[indices],
    "ingreso_uci": cohorte["ingreso_uci"].values[indices],
})

print(grande.shape)
print("Memoria:", round(grande.memory_usage(deep=True).sum() / 1024**2, 1), "MB")
```

```python
# El bucle fila a fila sobre un millón de filas tardaría demasiado para una
# clase, así que lo medimos sobre la DÉCIMA PARTE y extrapolamos.
sub = grande.head(100_000)

inicio = time.perf_counter()
suma = 0
for i in range(len(sub)):
    suma += sub["edad"].iloc[i]
media_bucle_g = suma / len(sub)
t_bucle_g = time.perf_counter() - inicio

inicio = time.perf_counter()
media_vect_g = sub["edad"].mean()
t_vect_g = time.perf_counter() - inicio

print(f"bucle sobre 100.000 filas       : {t_bucle_g:.3f} s")
print(f"vectorizado sobre 100.000 filas : {t_vect_g:.6f} s")
print(f"proporción: el bucle tarda {t_bucle_g/t_vect_g:.0f} veces más")
print()
print(f"las dos medias coinciden: {media_bucle_g:.4f} y {media_vect_g:.4f}")
print(f"extrapolando, el bucle sobre el millón entero tardaría ~{t_bucle_g*10:.0f} s")
```

```python
# Y sobre el millón entero, la versión vectorizada sigue siendo instantánea:
inicio = time.perf_counter()
print("media del millon:", round(grande["edad"].mean(), 4))
print(f"tiempo: {time.perf_counter() - inicio:.4f} s")
```

```python
# Un segundo experimento, para el error más caro que se comete al empezar:
# hacer crecer un objeto pegándole trozos dentro de un bucle.
valores = grande["edad"].tolist()[:20_000]

inicio = time.perf_counter()
acumulador = []
for v in valores:
    acumulador = acumulador + [v]      # crea una lista NUEVA y copia todo lo anterior
t_crece = time.perf_counter() - inicio

inicio = time.perf_counter()
acumulador2 = []
for v in valores:
    acumulador2.append(v)              # amplía la lista que ya existe
t_append = time.perf_counter() - inicio

print(f"lista recreada en cada vuelta : {t_crece:.3f} s")
print(f"lista ampliada con append()   : {t_append:.4f} s")
print(f"proporcion: {t_crece/t_append:.0f} veces mas lento")
```

```python
# Y ahora la parte que hay que ver con los propios ojos: repetimos la versión
# mala con el DOBLE de elementos. Si el coste creciera de forma proporcional,
# debería tardar el doble.
valores2 = grande["edad"].tolist()[:40_000]

inicio = time.perf_counter()
acumulador3 = []
for v in valores2:
    acumulador3 = acumulador3 + [v]
t_crece2 = time.perf_counter() - inicio

print(f"20.000 elementos : {t_crece:.3f} s")
print(f"40.000 elementos : {t_crece2:.3f} s")
print(f"al doblar los datos ha tardado {t_crece2/t_crece:.1f} veces más (no 2)")
```

```python
# Liberamos la memoria: 'grande' ocupa bastante y ya no lo necesitamos.
try:
    del grande, sub, valores, valores2
except NameError:
    pass   # si repites la celda, ya estaban borrados
import gc
gc.collect()
print("memoria liberada")
```

> **Interpretación de resultados**
>
> Los tiempos **exactos** dependen de la máquina (y en Colab, del servidor que te toque) y cambian de
> una ejecución a otra; lo que se mantiene es el **orden de magnitud**. Aproximadamente:
>
> | | bucle fila a fila | vectorizado | proporción |
> |---|---|---|---|
> | 1.800 filas | centésimas de segundo | décimas de milisegundo | de decenas a cientos de veces |
> | 100.000 filas | entre medio segundo y un par de segundos | menos de un milisegundo | de cientos a miles de veces |
> | 1.000.000 filas | varios segundos o más (extrapolado) | unos milisegundos | miles de veces |
>
> Fíjate en la última columna: la diferencia no es constante, **crece con el tamaño**. Y las dos
> medias son **idénticas** (unos 66,6 años). Mismo resultado, mismos datos, cientos o miles de veces
> más espera.
>
> El segundo experimento es todavía más claro: recrear la lista en cada vuelta tarda **cientos de
> veces** más que ampliarla con `append()` (aproximadamente 0,35 s frente a unas milésimas), y eso con solo 20.000
> elementos. Y al **doblar** los elementos no tarda el doble, sino aproximadamente 3-4 veces más: el coste crece
> de forma **cuadrática**. La causa es que `acumulador + [v]` no añade nada a la lista existente:
> crea una **nueva** y copia todo lo anterior. En la vuelta un millón copiaría un millón de valores.
> Por eso el ejemplo se queda en 40.000.
>
> **Por qué pasa esto.** `mean()` hace el cálculo en código compilado (C): recorre el millón de
> números de una sola pasada en un bloque de memoria contiguo (eso es lo que aporta numpy). El bucle
> hace un millón de viajes de ida y vuelta, y en cada uno Python comprueba tipos y construye objetos.
> La diferencia no es «Python es lento»: es que hay una forma de pedir las cosas que aprovecha cómo
> funciona la máquina.
>
> ---
>
> ### Y aquí empieza el Big Data
>
> Nuestra tabla de un millón de filas y **cuatro** columnas ocupa unos **30 MB**. Con las 21 columnas
> de la cohorte serían más de 300 MB. Un registro sanitario nacional (decenas de millones de
> pacientes, cientos de variables, años de seguimiento) se va a los **cientos de gigabytes**: no cabe
> en la memoria de tu portátil, y ninguna forma de escribir el bucle lo arregla.
>
> Cuando los datos dejan de caber en memoria hay tres salidas, y las tres están en el temario de la
> asignatura:
>
> 1. **No traer los datos**: dejarlos en una base de datos y pedirle a ella solo el resumen que
>    necesitas. Eso es **SQL**.
> 2. **Traerlos a trozos**: procesar por bloques y ir acumulando
>    (`pd.read_csv(..., chunksize=100_000)`).
> 3. **Repartirlos** entre varias máquinas que trabajan a la vez: la **computación distribuida**
>    (Spark, Hadoop) de la que habla el Big Data.
>
> Acabas de medir con un cronómetro por qué existen esas tres cosas.


## Bloque 8. Que el análisis se pueda repetir dentro de un año

*En la guía: «El cronómetro y la reproducibilidad».*

Un análisis **reproducible** es aquel que otra persona (o tú dentro de un año) puede volver a
ejecutar y obtener exactamente lo mismo. En investigación clínica esto no es una manía de
programadores: es lo que separa un resultado defendible de una anécdota.

Cinco costumbres, todas baratas:

1. **Un cuaderno o un script, no la consola.** Si solo existe en la consola, no existe.
2. **Orden fijo y siempre el mismo:** cabecera (qué hace, quién, cuándo) → importaciones → carga de
   datos → limpieza → análisis → resultados. (Este cuaderno de introducción no lo sigue del todo: está
   ordenado para explicar.)
3. **Rutas relativas** (Bloque 0).
4. **Semilla** para cualquier cosa con azar.
5. **Dejar constancia del entorno**: versiones de Python y de cada biblioteca.

Y una específica de los cuadernos, que es la causa nº 1 de resultados irreproducibles: **el orden en
que has ejecutado las celdas no tiene por qué ser el orden en que están escritas.** Antes de dar un
resultado por bueno, reinicia la sesión y ejecuta todo (menú `Entorno de ejecución`) y comprueba que sale lo
mismo de arriba abajo.


```python
# 4. Semilla: con la misma semilla, los mismos "números al azar".
rng = np.random.default_rng(42)
print(rng.integers(1, 100, 5))

rng = np.random.default_rng(42)
print(rng.integers(1, 100, 5))
```

```python
# 5. La "ficha técnica" del análisis.
import sys
print("Python     ", sys.version.split()[0])
print("pandas     ", pd.__version__)
print("numpy      ", np.__version__)
import matplotlib
print("matplotlib ", matplotlib.__version__)
```

> **Interpretación de resultados**
>
> Las dos llamadas con la **misma semilla** devuelven exactamente los mismos cinco números. Sin
> semilla saldrían distintos cada vez.
>
> Esto es lo que hace reproducible una partición de datos, una imputación o una validación cruzada:
> lo verás más adelante.
>
> El bloque de versiones es el equivalente de `sessionInfo()` en R. Pegarlo al final de un informe
> permite reproducir el resultado años después, cuando las versiones hayan cambiado.
